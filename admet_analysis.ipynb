```python
import pandas as pd
```

```python
path = "chosen_candidate_drugs/mir191"
# filename = pd.read_csv("chosen_candidate_drugs/swissadme-mir21.csv")
# filename = pd.read_csv("chosen_candidate_drugs/swissadme-mir96.csv")
# filename = pd.read_csv("chosen_candidate_drugs/swissadme-mir122.csv")
# filename = pd.read_csv("chosen_candidate_drugs/swissadme-mir155.csv")
filename = pd.read_csv("chosen_candidate_drugs/swissadme-mir191.csv")
```

```python
filename.columns
```

```python
import matplotlib.pyplot as plt
no_of_candidates_with_low_MW =len(filename[filename['MW']<500])
no_of_candidates_with_high_MW = len(filename[filename['MW']>500])
plt.figure(figsize=(6, 6))
bars = plt.bar(["MW<500","MW>500"],[no_of_candidates_with_low_MW,no_of_candidates_with_high_MW],align='center',width=0.2)
plt.title("Molecular weight of filtered candidates")

plt.title("Molecular weight of filtered candidates",fontsize=16, weight='bold')

plt.ylabel("Number of Molecules", fontsize=13)
plt.xlabel("Molecular weight", fontsize=13)
plt.xticks(fontsize=12)

plt.grid(axis='y', linestyle='--', alpha=0.6)

# remove top and right borders
plt.gca().spines['top'].set_visible(False)
plt.gca().spines['right'].set_visible(False)

# add value labels on bars
for bar in bars:
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2,
             height + 0.5,
             f'{height}',
             ha='center', va='bottom', fontsize=11)

plt.tight_layout()
plt.savefig(path+"_MW.png")
plt.show()
```

```python
# plt = filename['MW'].plot.hist()
# plt.set_xlabel("MW")
# plt.set_ylabel("No. of candidate drugs")
# plt.set_title("Molecular weight of filtered candidates targetting mir21")

import matplotlib.pyplot as plt
no_of_candidates_with_TPSA_accpt =len(filename.loc[(filename['TPSA']>20) & (filename['TPSA']<140)])

no_of_candidates_with_TPSA_rej = len(filename.loc[(filename['TPSA']<20) | 
                                                        (filename['TPSA']>140)])

plt.figure(figsize=(6, 6))
bars = plt.bar(["20<TPSA<140","TPSA>140"],
        [no_of_candidates_with_TPSA_accpt,no_of_candidates_with_TPSA_rej],
        align='center',width=0.2)
plt.title("Topological Polar Surface Area",fontsize=16, weight='bold')
plt.ylabel("Number of Molecules", fontsize=13)
plt.xlabel("TPSA class", fontsize=13)
plt.xticks(fontsize=12)

plt.grid(axis='y', linestyle='--', alpha=0.6)

# remove top and right borders
plt.gca().spines['top'].set_visible(False)
plt.gca().spines['right'].set_visible(False)

# add value labels on bars
for bar in bars:
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2,
             height + 0.5,
             f'{height}',
             ha='center', va='bottom', fontsize=11)

plt.tight_layout()
plt.savefig(path+"_TPSA.png")
plt.show()
```
